# Membership inference by confidence thresholding (toy data)

> **Scope.** Run this only against models and apps you own or the toy targets defined here. No third-party systems, no real people's data. Everything in this repository is synthetic.

A deliberately overfit random forest on synthetic data, and the simplest membership test: guess "member" when the model's confidence passes a threshold (the idea behind Shokri et al., 2017, "Membership Inference Attacks Against Machine Learning Models", IEEE S&P).

```python
%pip install -q scikit-learn
```

```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.ensemble import RandomForestClassifier

X, y = make_classification(n_samples=2000, n_features=6, n_informative=4, flip_y=0.15, random_state=0)
X_in, y_in, X_out = X[:1000], y[:1000], X[1000:]
model = RandomForestClassifier(n_estimators=100, max_depth=None, bootstrap=False, random_state=0).fit(X_in, y_in)

conf_in = model.predict_proba(X_in).max(1)
conf_out = model.predict_proba(X_out).max(1)
for t in (0.8, 0.9, 0.95, 0.99, 1.0):
    acc = ((conf_in >= t).sum() + (conf_out < t).sum()) / (len(conf_in) + len(conf_out))
    print(f"threshold {t:.2f}: attack accuracy {acc:.3f} (chance = 0.500)")
```
