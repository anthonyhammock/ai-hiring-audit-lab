```python
import hashlib, urllib.request, pandas as pd
base = "https://raw.githubusercontent.com/anthonyhammock/ai-hiring-audit-lab/main/m21-fairness-dataset/data/"
sums = dict(line.split()[::-1] for line in urllib.request.urlopen(base + "SHA256SUMS").read().decode().splitlines())
for name in ["applicants.csv", "vendor_summary.csv", "external_benchmark.csv"]:
    data = urllib.request.urlopen(base + name).read()
    open(name, "wb").write(data)
    print(name, "hash OK" if hashlib.sha256(data).hexdigest() == sums[name] else "HASH MISMATCH")
```
